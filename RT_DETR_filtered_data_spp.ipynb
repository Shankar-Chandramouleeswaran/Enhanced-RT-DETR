```python
import os
import random
import json
from pycocotools.coco import COCO
from collections import defaultdict
from tqdm import tqdm

# Paths
ann_file = '/home/exouser/Downloads/SPP/coco/annotations/instances_train2017.json'
save_path = '/home/exouser/Downloads/SPP/coco/annotations/instances_train2017_subset.json'

# Number of images per category to sample
images_per_category = 700

# Target category IDs
target_categories = [1, 2, 3, 4, 5, 6, 8, 10, 11, 13, 14]  # as per your list

# Load COCO annotations
coco = COCO(ann_file)

# Category name mapping (optional)
cat_id_to_name = {cat['id']: cat['name'] for cat in coco.loadCats(target_categories)}

# Store image ids to keep
selected_image_ids = set()

# Sample images for each category
for cat_id in target_categories:
    img_ids = coco.getImgIds(catIds=[cat_id])
    sampled = random.sample(img_ids, min(images_per_category, len(img_ids)))
    selected_image_ids.update(sampled)
    print(f"Category '{cat_id_to_name[cat_id]}' ({cat_id}): selected {len(sampled)} images")

# Filter images
selected_image_ids = list(selected_image_ids)
all_images = coco.loadImgs(selected_image_ids)

# Filter annotations
selected_annots = []
image_id_set = set(selected_image_ids)

for ann in tqdm(coco.dataset['annotations']):
    if ann['image_id'] in image_id_set and ann['category_id'] in target_categories:
        selected_annots.append(ann)

# Filter categories
selected_cats = [cat for cat in coco.dataset['categories'] if cat['id'] in target_categories]

# Save new JSON
filtered_dataset = {
    "images": all_images,
    "annotations": selected_annots,
    "categories": selected_cats
}

with open(save_path, 'w') as f:
    json.dump(filtered_dataset, f)

print(f"\n Saved filtered dataset with {len(all_images)} images and {len(selected_annots)} annotations to:\n{save_path}")
```

```python
import json
import os
import random
from collections import defaultdict
from tqdm import tqdm

# Settings
categories_to_keep = [1, 2, 3, 4, 5, 6, 8, 10, 11, 13, 14]
max_per_category = 700  # or fewer if you want
val_ann_path = "/home/exouser/Downloads/SPP/coco/annotations/instances_val2017.json"
val_img_dir = "/home/exouser/Downloads/SPP/coco/val2017"
save_ann_path = "/home/exouser/Downloads/SPP/coco/annotations/instances_val2017_subset.json"

# Load data
with open(val_ann_path, "r") as f:
    coco = json.load(f)

# Collect image-ids per category
cat_to_imgs = defaultdict(set)
imgid_to_anns = defaultdict(list)

for ann in coco["annotations"]:
    if ann["category_id"] in categories_to_keep:
        cat_to_imgs[ann["category_id"]].add(ann["image_id"])
        imgid_to_anns[ann["image_id"]].append(ann)

# Randomly pick images per category
selected_img_ids = set()
for cat_id in categories_to_keep:
    imgs = list(cat_to_imgs[cat_id])
    selected = random.sample(imgs, min(max_per_category, len(imgs)))
    selected_img_ids.update(selected)

# Filter images
selected_imgs = [img for img in coco["images"] if img["id"] in selected_img_ids]

# Filter annotations
selected_anns = [ann for ann in coco["annotations"] if ann["image_id"] in selected_img_ids and ann["category_id"] in categories_to_keep]

# Filter categories
selected_cats = [cat for cat in coco["categories"] if cat["id"] in categories_to_keep]

# Save new annotation JSON
new_coco = {
    "images": selected_imgs,
    "annotations": selected_anns,
    "categories": selected_cats
}

with open(save_ann_path, "w") as f:
    json.dump(new_coco, f)

print(f" Saved: {save_ann_path} with {len(selected_imgs)} images and {len(selected_anns)} annotations.")
```

```python
import os
import json

def verify_coco_dataset(img_dir, ann_path, expected_cat_ids):
    print(f" Verifying dataset:\n - Images: {img_dir}\n - Annotations: {ann_path}\n")

    with open(ann_path, "r") as f:
        data = json.load(f)

    images = data.get("images", [])
    annotations = data.get("annotations", [])
    categories = data.get("categories", [])

    image_files = set(os.listdir(img_dir))
    image_names_in_json = set(img["file_name"] for img in images)
    image_ids_in_json = set(img["id"] for img in images)

    # ? Check if all image files mentioned in JSON exist
    missing_images = [img for img in image_names_in_json if img not in image_files]
    print(f" Total images in JSON: {len(image_names_in_json)}")
    print(f" Total images in folder: {len(image_files)}")
    print(f" Missing images: {len(missing_images)}")
    if missing_images:
        print(" Missing image filenames:", missing_images[:5])  # limit print

    # ? Check if all annotation image_ids exist in image list
    invalid_ids = [ann['image_id'] for ann in annotations if ann['image_id'] not in image_ids_in_json]
    print(f" Total annotations: {len(annotations)}")
    print(f" Annotations with invalid image_id: {len(invalid_ids)}")

    # ? Check category IDs
    invalid_cats = [ann['category_id'] for ann in annotations if ann['category_id'] not in expected_cat_ids]
    print(f" Expected category IDs: {sorted(expected_cat_ids)}")
    print(f" Annotations with invalid category_id: {len(invalid_cats)}")

    # ? Print categories
    cat_names = {cat["id"]: cat["name"] for cat in categories}
    print(f" Categories in annotation file ({len(cat_names)}): {cat_names}")

    print("\n? Check completed.\n")


# Set paths
train_img_dir = '/home/exouser/Downloads/SPP/coco/train2017_filtered/'
train_ann = '//home/exouser/Downloads/SPP/coco/annotations/instances_train2017_subset_remapped.json'

val_img_dir = '/home/exouser/Downloads/SPP/coco/val2017_filtered/'
val_ann = '/home/exouser/Downloads/SPP/coco/annotations/instances_val2017_subset_remapped.json'

# Expected remapped category IDs
expected_ids = set(range(11))  # 0 to 10

# Run checks
verify_coco_dataset(train_img_dir, train_ann, expected_ids)
verify_coco_dataset(val_img_dir, val_ann, expected_ids)
```

```python
import os
import json
import matplotlib.pyplot as plt
import pandas as pd

log_path = "/home/exouser/Downloads/SPP/SPPv2/rtdetr_pytorch/output/rtdetr_r50vd_6x_coco/log.txt"

# Containers
epoch_losses = []
ap_metrics = {
    "AP": [], "AP50": [], "AP75": [],
    "APS": [], "APM": [], "APL": [],
}
epochs_recorded = set()

# Read and parse the log
with open(log_path, "r") as f:
    for line in f:
        try:
            log_entry = json.loads(line.strip())
        except json.JSONDecodeError:
            continue

        epoch = log_entry.get("epoch")
        if epoch is None or epoch in epochs_recorded:
            continue  # Avoid duplicate or missing epochs

        epochs_recorded.add(epoch)

        # Extract loss
        loss = log_entry.get("train_loss")
        if loss is not None:
            epoch_losses.append((epoch, loss))

        # Extract AP metrics (ignore -1.0 or missing)
        ap_list = log_entry.get("test_coco_eval_bbox")
        if ap_list and len(ap_list) == 12:
            ap_val = ap_list[0]
            ap50 = ap_list[1]
            ap75 = ap_list[2]
            aps = ap_list[3]
            apm = ap_list[4]
            apl = ap_list[5]

            for key, value in zip(["AP", "AP50", "AP75", "APS", "APM", "APL"],
                                  [ap_val, ap50, ap75, aps, apm, apl]):
                if value is not None and value >= 0:
                    ap_metrics[key].append((epoch, value * 100))  # convert to percentage

# Sort by epoch
epoch_losses.sort()
for key in ap_metrics:
    ap_metrics[key].sort()

# ----------- ?? Plotting Training Loss --------------
loss_df = pd.DataFrame(epoch_losses, columns=["Epoch", "Loss"])
plt.figure(figsize=(10, 5))
plt.plot(loss_df["Epoch"], loss_df["Loss"], marker='o', label="Training Loss", color="red")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss vs Epoch")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# ----------- ?? Plotting AP Metrics ----------------
plt.figure(figsize=(12, 6))
for key, data in ap_metrics.items():
    if not data:
        continue
    df = pd.DataFrame(data, columns=["Epoch", "AP"])
    plt.plot(df["Epoch"], df["AP"], marker='o', label=key)

plt.xlabel("Epoch")
plt.ylabel("AP (%)")
plt.title("AP Metrics vs Epoch")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# ----------- ?? Find Best Epochs per AP -------------
best_scores = []
full_names = {
    "AP":   "Average Precision  (AP) @[ IoU=0.50:0.95 | area=   all ]",
    "AP50": "Average Precision  (AP) @[ IoU=0.50      | area=   all ]",
    "AP75": "Average Precision  (AP) @[ IoU=0.75      | area=   all ]",
    "APS":  "Average Precision  (AP) @[ IoU=0.50:0.95 | area= small ]",
    "APM":  "Average Precision  (AP) @[ IoU=0.50:0.95 | area=medium ]",
    "APL":  "Average Precision  (AP) @[ IoU=0.50:0.95 | area= large ]",
}

for key, data in ap_metrics.items():
    if not data:
        continue
    df = pd.DataFrame(data, columns=["Epoch", "AP"])
    best_row = df.loc[df["AP"].idxmax()]
    best_scores.append((
        full_names[key],
        int(best_row["Epoch"]),
        round(best_row["AP"], 2)
    ))

# Create DataFrame for display
best_df = pd.DataFrame(best_scores, columns=["Metric Description", "Best Epoch", "Best AP (%)"])

# Display cleanly
print("\nBest Epochs for Each AP Metric:\n")
print(best_df.to_string(index=False))

```
