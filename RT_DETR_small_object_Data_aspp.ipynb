```python
import json
import os

# Define paths
coco_annotation_path = 'coco1/annotations/instances_train2017.json'
filtered_annotation_path = 'coco1/annotations/instances_train2017_small.json'
SMALL_OBJECT_THRESHOLD = 32 * 32  # 1024 px²

# Load COCO Annotations
with open(coco_annotation_path, 'r') as f:
    coco_data = json.load(f)

small_annotations = []
small_image_ids = set()

# Filter annotations with small objects
for ann in coco_data['annotations']:
    area = ann["bbox"][2] * ann["bbox"][3]  # width * height
    if area < SMALL_OBJECT_THRESHOLD:
        small_annotations.append(ann)
        small_image_ids.add(ann["image_id"])

# Filter images that contain small objects
filtered_images = [img for img in coco_data['images'] if img['id'] in small_image_ids]

# Save new dataset
filtered_coco_data = {
    "info": coco_data.get("info", {}),
    "licenses": coco_data.get("licenses", []),
    "images": filtered_images,
    "annotations": small_annotations,
    "categories": coco_data.get("categories", [])
}

with open(filtered_annotation_path, 'w') as f:
    json.dump(filtered_coco_data, f)

print(f"Filtered dataset saved at {filtered_annotation_path}")
```

```python
import os
import cv2
import numpy as np
import albumentations as A
from albumentations.pytorch import ToTensorV2
import json

# ✅ Fixed transform using 'height' and 'width' instead of 'size'
transform = A.Compose([
    A.RandomResizedCrop(height=640, width=640, scale=(0.5, 1.0), ratio=(0.75, 1.333), 
                         interpolation=cv2.INTER_LINEAR, p=1.0),
    A.HorizontalFlip(p=0.5),
    A.VerticalFlip(p=0.5),
    A.RandomBrightnessContrast(p=0.2),
    A.GaussNoise(var_limit=(10, 50), p=0.3),
    ToTensorV2()
])

# Path to your filtered annotations (make sure filtered_annotation_path is defined)
filtered_annotation_path = 'coco1/annotations/instances_train2017_small.json'

# Load filtered dataset
with open(filtered_annotation_path, 'r') as f:
    filtered_coco_data = json.load(f)

image_dir = "coco1/train2017/"
augmented_dir = "coco1/train2017_augmented/"
os.makedirs(augmented_dir, exist_ok=True)

for img_info in filtered_coco_data['images']:
    img_path = os.path.join(image_dir, img_info['file_name'])
    image = cv2.imread(img_path)
    if image is None:
        print(f"Warning: Image {img_info['file_name']} not found.")
        continue

    # Convert BGR to RGB
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Apply augmentation transform
    augmented = transform(image=image)
    aug_img = augmented['image']

    # Convert the PyTorch tensor to a NumPy array and permute dimensions from (C, H, W) to (H, W, C)
    aug_img = aug_img.permute(1, 2, 0).numpy()

    aug_img_path = os.path.join(augmented_dir, img_info['file_name'])
    # Convert RGB back to BGR for cv2.imwrite
    cv2.imwrite(aug_img_path, cv2.cvtColor(aug_img, cv2.COLOR_RGB2BGR))

print(f"✅ Augmented images saved in {augmented_dir}")
```

```python
import json
import os
import shutil

# ✅ Define paths
original_json_path = "/xdisk/ldbrown/maheshg/mahesh/coco1/annotations/instances_train2017_small.json" 
filtered_json_path = "/xdisk/ldbrown/maheshg/mahesh/coco1/annotations/instances_train2017_small_selected.json" 
image_dir = "/xdisk/ldbrown/maheshg/mahesh/coco1/train2017_augmented/" 
filtered_image_dir = "/xdisk/ldbrown/maheshg/mahesh/coco1/train2017_augmented_selected/" 

# ✅ Ensure the output folder exists
os.makedirs(filtered_image_dir, exist_ok=True)

# ✅ Selected category IDs from COCO (original category IDs)
selected_category_ids = [1, 2, 3, 4, 5, 6, 7, 8, 10, 11, 13, 14, 15, 28, 31, 32, 33, 34]

# ✅ Create a mapping from original COCO category IDs → new category IDs (0 to num_classes-1)
category_id_mapping = {orig_id: new_id for new_id, orig_id in enumerate(selected_category_ids)}

# ✅ Load original COCO annotations
with open(original_json_path, "r") as f:
    coco_data = json.load(f)

# ✅ Filter and remap annotations for selected categories
filtered_annotations = []
for ann in coco_data["annotations"]:
    if ann["category_id"] in selected_category_ids:
        ann["category_id"] = category_id_mapping[ann["category_id"]]  # Remap category ID
        filtered_annotations.append(ann)

# ✅ Collect image IDs that have the selected categories
selected_image_ids = {ann["image_id"] for ann in filtered_annotations}

# ✅ Filter images that contain selected objects
filtered_images = [img for img in coco_data["images"] if img["id"] in selected_image_ids]

# ✅ Copy selected images to new folder
for img_info in filtered_images:
    src_path = os.path.join(image_dir, img_info["file_name"])
    dst_path = os.path.join(filtered_image_dir, img_info["file_name"])
    
    if os.path.exists(src_path):  # Ensure image exists before copying
        shutil.copy(src_path, dst_path)

# ✅ Save the new filtered dataset with remapped categories
filtered_coco_data = {
    "info": coco_data.get("info", {}),
    "licenses": coco_data.get("licenses", []),
    "images": filtered_images,
    "annotations": filtered_annotations,
    "categories": [
        {"id": new_id, "name": cat["name"], "supercategory": cat["supercategory"]}
        for new_id, cat in enumerate(coco_data["categories"])
        if cat["id"] in selected_category_ids
    ]
}

with open(filtered_json_path, "w") as f:
    json.dump(filtered_coco_data, f, indent=4)

print(f"✅ Filtered dataset saved: {filtered_json_path}")
print(f"✅ {len(filtered_images)} images copied to: {filtered_image_dir}")
print(f"✅ Category IDs have been remapped to a contiguous range (0-{len(selected_category_ids)-1}).")
```
